# PUL7SAR Phase 18 — Semi-Automatic Colab GPU Runner

This notebook keeps GitHub `phase18/story-intelligence` as the source of truth and reduces the Golden Visual GPU loop to three controlled cells. It never checks out or updates `main`. Generation success remains `publication_ready=false` until semantic and Golden-quality review pass.

```python
from pathlib import Path
import os, subprocess

ROOT = Path('/content/pul7sar-bot')
BRANCH = 'phase18/story-intelligence'
REPO = 'https://github.com/pulsar7official/pul7sar-bot.git'

if not ROOT.exists():
    subprocess.run(['git', 'clone', '--branch', BRANCH, '--single-branch', REPO, str(ROOT)], check=True)
os.chdir(ROOT)
current = subprocess.check_output(['git', 'branch', '--show-current'], text=True).strip()
assert current == BRANCH, f'Refusing to run outside {BRANCH}: {current}'
subprocess.run(['git', 'pull', '--ff-only', 'origin', BRANCH], check=True)
print('Branch:', current)
print('HEAD:', subprocess.check_output(['git', 'rev-parse', '--short', 'HEAD'], text=True).strip())
subprocess.run(['nvidia-smi'], check=True)
```
